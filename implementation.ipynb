```python
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras.preprocessing.image import ImageDataGenerator
```

```python
dataset_path="C://sushmitha one drive//Documents//66B8//CNN//PetImages"
```

```python
#Image Processing
datagen=ImageDataGenerator(
    #Normalize pixels
    rescale=1./255,
    #Data augmentation
    shear_range=0.2,
    zoom_range=0.2,
    horizontal_flip=True,

    #split dataset
    validation_split=0.2 #validation split used to create train and validation data
)
```

```python
#Loading the training data
train_data=datagen.flow_from_directory(
    dataset_path,
    target_size=(128,128),
    batch_size=32,
    class_mode='binary',
    subset='training'
 )
```

```python
# Load validation data 

val_data = datagen.flow_from_directory(
    dataset_path,
    target_size = (128, 128),  # Resize images to 150x150
    batch_size = 32,
    class_mode = 'binary',  # Binary classification (cat vs dog)
    subset = 'validation'  # Use the validation subset
)
```

```python
print(train_data.class_indices)
```

```python
# Building CNN  model
model=keras.Sequential()
```

```python
#First Convolution Layer


model.add(keras.layers.Conv2D(
    filters=32,
    kernel_size=(3,3),
    activation='relu',
    input_shape=(128,128,3)
))
```

```python
#First Pooling layer

model.add(
    keras.layers.MaxPooling2D(
        pool_size=(2,2)
    )
)
```

```python
# 2 Conv Layer

model.add(keras.layers.Conv2D(
    filters=32,
    kernel_size=(3,3),
    activation='relu'
))
```

```python
#Second Pooling layer
model.add(
    keras.layers.MaxPooling2D(
        pool_size=(2,2)
    )
)
```

```python
#Flattern

model.add(keras.layers.Flatten())
```

```python
#Dense Layer
model.add(
    keras.layers.Dense(
        units=128,
        activation='relu'
    )
)
```

```python
#output layer
model.add(keras.layers.Dense(
    units=1,
    activation='sigmoid'
))
```

```python
#View Model architecture

model.summary()
```

```python
# Compile CNN
model.compile(optimizer='adam',
              loss='binary_crossentropy',
              metrics=['accuracy'])
```

```python
# Train
history=model.fit(
    train_data,
    epochs=10,
    validation_data=val_data
)
```

```python
model.save("cnn_pet_classifier.h5")
```

```python
#Evaluate the model
loss,accuracy=model.evaluate(val_data)
print("Accuracy:",accuracy)
```
